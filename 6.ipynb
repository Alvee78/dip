```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Read two grayscale images
image1 = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)
image2 = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)
```

```python
# -----------------------------
# Arithmetic Operations
# -----------------------------

# Addition
addition = cv2.add(image1, image2)

# Subtraction
subtraction = cv2.subtract(image1, image2)
```

```python
# -----------------------------
# Logical Operations
# -----------------------------

# AND
logical_and = cv2.bitwise_and(image1, image2)

# OR
logical_or = cv2.bitwise_or(image1, image2)

# NOT
logical_not = cv2.bitwise_not(image1)
```

```python

# -----------------------------
# Display Results
# -----------------------------

plt.figure(figsize=(12, 8))

plt.subplot(2, 4, 1)
plt.imshow(image1, cmap="gray")
plt.title("Image 1")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(image2, cmap="gray")
plt.title("Image 2")
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(addition, cmap="gray")
plt.title("Addition")
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(subtraction, cmap="gray")
plt.title("Subtraction")
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(logical_and, cmap="gray")
plt.title("AND")
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(logical_or, cmap="gray")
plt.title("OR")
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(logical_not, cmap="gray")
plt.title("NOT")
plt.axis("off")

plt.tight_layout()
plt.show()
```
