```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Read image in grayscale
image = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python
# =================================
# 1. Contrast Stretching
# =================================

min_val = np.min(image)
max_val = np.max(image)

contrast = ((image - min_val) / (max_val - min_val) * 255).astype(np.uint8)
```

```python
# =================================
# 2. Bit-Plane Slicing
# =================================

bit_planes = []

for i in range(8):
    # Extract bit i
    bit_plane = ((image >> i) & 1) * 255
    bit_planes.append(bit_plane)
```

```python

# =================================
# Display Contrast Stretching
# =================================

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(contrast, cmap="gray")
plt.title("Contrast Stretched")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# =================================
# Display Bit Planes
# =================================

plt.figure(figsize=(12, 8))

for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_planes[i], cmap="gray")
    plt.title("Bit Plane " + str(i))
    plt.axis("off")

plt.tight_layout()
plt.show()
```
