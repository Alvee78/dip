```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Read image
image = cv2.imread("image.jpeg")
```

```python
# Convert BGR to RGB for displaying
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python

# Get image size
height, width = image.shape[:2]
```

```python
# -----------------------------
# 1. Resizing / Zooming
# -----------------------------
resized = cv2.resize(image, (width * 2, height * 2))
```

```python
# -----------------------------
# 2. Rotation
# -----------------------------
center = (width // 2, height // 2)

rotation_matrix = cv2.getRotationMatrix2D(
    center,
    45,       # Rotation angle
    1         # Scale
)

rotated = cv2.warpAffine(
    image,
    rotation_matrix,
    (width, height)
)
```

```python
# -----------------------------
# 3. Translation
# -----------------------------
translation_matrix = cv2.getRotationMatrix2D(
    (0, 0),
    0,
    1
)

translation_matrix[0, 2] = 100  # Move right
translation_matrix[1, 2] = 50   # Move down

translated = cv2.warpAffine(
    image,
    translation_matrix,
    (width, height)
)
```

```python
# -----------------------------
# Display Results
# -----------------------------
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(resized)
plt.title("Resized / Zoomed")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(rotated)
plt.title("Rotated 45°")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(translated)
plt.title("Translated")
plt.axis("off")

plt.tight_layout()
plt.show()
```
