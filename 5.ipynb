```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Read color image
image = cv2.imread("image.jpeg")
```

```python
# Convert BGR to RGB
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python

# --------------------------------
# 1. Color Image → Grayscale Image
# --------------------------------

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
# --------------------------------
# 2. Color Image → Binary Image
# --------------------------------

_, binary = cv2.threshold(
    gray,
    127,
    255,
    cv2.THRESH_BINARY
)
```

```python

# --------------------------------
# 3. Separate RGB into R, G, B planes
# --------------------------------

R = rgb[:, :, 0]
G = rgb[:, :, 1]
B = rgb[:, :, 2]
```

```python

# --------------------------------
# 4. Combine R, G, B planes
# --------------------------------

combined = cv2.merge([R, G, B])
```

```python

# --------------------------------
# Display Results
# --------------------------------

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(rgb)
plt.title("Original Color Image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(R, cmap="gray")
plt.title("Red Plane")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(G, cmap="gray")
plt.title("Green Plane")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(B, cmap="gray")
plt.title("Blue Plane")
plt.axis("off")

plt.tight_layout()
plt.show()


# Display reconstructed image
plt.figure(figsize=(5, 5))
plt.imshow(combined)
plt.title("Combined RGB Image")
plt.axis("off")
plt.show()
```
