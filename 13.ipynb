```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read image in grayscale
image = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python
# Create output images
median = np.zeros_like(image)
maximum = np.zeros_like(image)
minimum = np.zeros_like(image)
```

```python
# Apply 3x3 filters
for i in range(1, image.shape[0] - 1):
    for j in range(1, image.shape[1] - 1):

        # Get 3x3 neighborhood
        neighborhood = image[i-1:i+2, j-1:j+2]

        # Median filter
        median[i, j] = np.median(neighborhood)

        # Max filter
        maximum[i, j] = np.max(neighborhood)

        # Min filter
        minimum[i, j] = np.min(neighborhood)
```

```python
# Display results
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(median, cmap="gray")
plt.title("Median Filter")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(maximum, cmap="gray")
plt.title("Max Filter")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(minimum, cmap="gray")
plt.title("Min Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```
