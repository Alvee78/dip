```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
image = cv2.imread("image.jpeg", 0)
```

```python
# =================================
# 1. Thresholding
# =================================

threshold = 127

binary = np.zeros_like(image)

binary[image > threshold] = 255
```

```python
# =================================
# 2. Region-Based Segmentation
#    Simple Region Growing
# =================================

segmented = np.zeros_like(image)

# Starting point (seed)
seed_x = image.shape[0] // 2
seed_y = image.shape[1] // 2

# Region intensity range
lower = image[seed_x, seed_y] - 20
upper = image[seed_x, seed_y] + 20

# Select pixels belonging to the region
segmented[
    (image >= lower) & (image <= upper)
] = 255
```

```python
# =================================
# Display Results
# =================================

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(binary, cmap="gray")
plt.title("Thresholding")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(segmented, cmap="gray")
plt.title("Region Segmentation")
plt.axis("off")

plt.tight_layout()
plt.show()
```
