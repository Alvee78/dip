```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Read image in grayscale
image = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python
# Set intensity range
low = 100
high = 200
```

```python

# --------------------------------
# 1. Gray-Level Slicing
#    Without Background
# --------------------------------

without_background = image.copy()

without_background[(image >= low) & (image <= high)] = 255
without_background[(image < low) | (image > high)] = 0
```

```python
# --------------------------------
# 2. Gray-Level Slicing
#    With Background
# --------------------------------

with_background = image.copy()

with_background[(image >= low) & (image <= high)] = 255
```

```python
# --------------------------------
# Display Results
# --------------------------------

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(without_background, cmap="gray")
plt.title("Without Background")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(with_background, cmap="gray")
plt.title("With Background")
plt.axis("off")

plt.tight_layout()
plt.show()
```
