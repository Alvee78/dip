```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read image in grayscale
image = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python
# Create output image
smooth = np.zeros_like(image)

# 3 x 3 averaging filter
kernel_size = 3
```

```python
# Apply averaging filter
for i in range(1, image.shape[0] - 1):
    for j in range(1, image.shape[1] - 1):

        # Take 3 x 3 neighborhood
        neighborhood = image[i-1:i+2, j-1:j+2]

        # Calculate average
        average = np.sum(neighborhood) / 9

        # Store result
        smooth[i, j] = int(average)
```

```python
# Display images
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(smooth, cmap="gray")
plt.title("3 × 3 Averaging Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```
