```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read image in grayscale
image = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
image.shape
```

```python
# -----------------------------
# Spatial Resolution
# -----------------------------

spatial_128 = cv2.resize(image, (128, 128))
spatial_64 = cv2.resize(image, (64, 64))
spatial_32 = cv2.resize(image, (32, 32))
```

```python
# -----------------------------
# Intensity Resolution
# -----------------------------

# 256 intensity levels
intensity_256 = image

# 16 intensity levels
intensity_16 = np.floor(image / 16) * 16

# 4 intensity levels
intensity_4 = np.floor(image / 64) * 64
```

```python
# -----------------------------
# Display Spatial Resolution
# -----------------------------

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(spatial_128, cmap="gray")
plt.title("128 x 128")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(spatial_64, cmap="gray")
plt.title("64 x 64")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(spatial_32, cmap="gray")
plt.title("32 x 32")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# Display Intensity Resolution
# -----------------------------

plt.figure(figsize=(10, 6))

plt.subplot(1, 3, 1)
plt.imshow(intensity_256, cmap="gray")
plt.title("256 Levels")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(intensity_16, cmap="gray")
plt.title("16 Levels")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(intensity_4, cmap="gray")
plt.title("4 Levels")
plt.axis("off")

plt.tight_layout()
plt.show()
```
